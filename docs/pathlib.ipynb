## Playing with files

```python
my_file = cwd / "my_file.txt"
print(my_file, my_file.exists())
```

```python
my_file.write_text("12")
```

```python
my_file.exists()
```

```python
my_file.read_text()
```

```python
my_file.suffix
```

```python
my_file.with_suffix(".py")
```

```python
my_file.with_name("something_else")
```

```python
with my_file.open("r") as f:
    print(f.read())
```

```python
my_file.unlink()
```

```python
my_file.exists()
```
