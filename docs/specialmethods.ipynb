# Some special methods

## `__len__`, `__call__`, `__getitem__`

```python
class MyClass:
    
    def __len__(self) -> int:
        return 12
    
    def __getitem__(self, ind):
        print(f"Calling __getitem__ with {ind}")
        return ind
    
    def __call__(self, *args, **kwargs):
        print(f"Calling __call__ with {args} and {kwargs}")
        return args, kwargs
    
inst = MyClass()
```

```python
len(inst)
```

```python
inst(2, x=12)
```

```python
inst[3]
```

```python
inst[:3]
```

```python
inst[1:3]
```

```python
inst[slice(1, 3, 2)]
```

```python
s = slice(1, 20, 2)
print(s.start, s.stop, s.step)
```

```python
s.indices(12)
```

**Note**: for a dict-like object, you should probably implement `__setitem__`, `__delitem__`, `__contains__`...

## `__repr__`, `__add__`, `__radd__`

```python
class MyString:
    
    def __init__(self, string: str):
        self.string = string

print(MyString("test"))
```

```python
class MyString:
    
    def __init__(self, string: str) -> None:
        self.string = string
    
    def __repr__(self) -> str:
        return "whatever"
        return f"{self.__class__.__name__}({self.string!r})"
    
    def __str__(self) -> str:
        return "str method"
    
print(MyString("test"))
```

```python
f'{MyString("tests")!r}'
```

Note: there is also `__str__` (which defaults to using `__repr__`). f-string use format with `str` by default, but you can use `{inst!r}` to use `repr`.

```python
print(str("test\n"))
```

```python
repr("test\n")
```

```python
class MyString2(MyString):
    
    def __add__(self, other: int):
        print(f"adding {other}")
        return self.__class__(self.string + str(other))

MyString2("Hello") + 18
```

```python
12 + MyString2("Hello")  # you'd need to implement __radd__
```

**Note**: for your objects to be sortable, the following could be useful: `__eq__`, `__lt__` (lower than) etc...

## `__init_subclass__`

```python
CALLBACK_CLASSES = {}

class BaseCallback:
    
    @classmethod
    def __init_subclass__(cls) -> None:
        if cls.__name__ in CALLBACK_CLASSES:
            print(f"Redefining class: {cls.__name__}")
        CALLBACK_CLASSES[cls.__name__] = cls
    
    def call(self, *args, **kwargs) -> None:
        raise NotImplementedError
        
```

```python
CALLBACK_CLASSES
```

```python
class PrinterCallback(BaseCallback):
    
    def call(self, *args, **kwargs) -> None:
        print(args, kwargs)

class DoNothingCallback(BaseCallback):
    
    def call(self, *args, **kwargs) -> None:
        pass
```

```python
CALLBACK_CLASSES  # stores all the implementations in one place :)
```

```python
cb = PrinterCallback()
```
