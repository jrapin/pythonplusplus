# Mutable & Immutable

## Example: list and tuples

```python
my_tuple = (1, 2)
my_list = [1, 2]
```

```python
my_list[1] = 3
```

```python
my_tuple[1] = 3
```

```python
hash(my_tuple)
```

```python
hash(my_list)
```

```python
d = {}
d[my_tuple] = 12
```

```python
d[my_list]
```

**Take home message**: mutables (like lists) can be modified, unlike immutables. One consequence is that they can't be hashed (since they can be modified) and therefore cannot be used as dict keys.

## Mutables and immutables

**mutables**: list, dict, set, most objects in general

**immutables**: int, float, bool, str, tuples, frozenset

```python
x = [1, 2]
id(x)
```

```python
x += [3]
id(x)
```

```python
string = "ab"
id(string)
```

```python
string += "c"
id(string)
```

**Note**: a consequence of strings immutability is that operations on strings create new objects, and can therefore be slow on long strings.

## Gotchas

**Never ever use a mutable as default variables!**

```python
def my_function(a_list=[]):
    a_list.append(len(a_list))
    return a_list
```

```python
my_function([1, 2])
```

```python
my_function([1, 2])
```

```python
my_function()
```

```python
my_function()
```

Be cautious with references of mutable objects

```python
my_lists = 3 * [[0]]
print(my_lists)
```

```python
my_lists[0][0] = 12
print(my_lists)
```

## My own immutable class
Just to be clear: **you most probably don't need this**

### Slots

```python
class MyClass:
    __slots__ = ["a", "b"]
    
    def __init__(self):
        self.a = 12
        self.b = "a string"
        
inst = MyClass()
```

```python
inst.c = 12
```

```python
inst.__dict__  # attributes are not stored in a the dict
```

```python
inst.a = 13  # that works, so it's still mutable
```

### `__setattr__`

**Note**: don't do it either

```python
class MyClass:
    
    def __init__(self):
        # how ugly is that?!
        self.a: int
        self.b: str
        self.__dict__.update(a=12, b="a string")
        # why? because since we cannot set attributes, we cannot
        # initialize them either in the init...
    
    def __setattr__(self, name, value) -> None:
        raise RuntimeError("This class is not mutable")
        
inst = MyClass()
```

```python
inst.a
```

```python
inst.a = 13
```

```python
del inst.a
inst.a  # actually we also need to override __delattr__
```
