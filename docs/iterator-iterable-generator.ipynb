```python
class Iterable:
    def __init__(self):
        self.k = 0
    
    def __next__(self):
        self.k += 1
        return self.k
    
    def __iter__(self):
        return self
```

# Definitions

**iterable**: an object with an `__iter__` method, providing an iterator (eg: sequences, like list, tuples, array). Sources of elements in for-loops are *always iterables*.

**iterator**: an object that implements `__next__` to return a new item, and raises `StopIteration` when there is no item left. Most times, iterators are iterable, because they have an `__iter__` method which returns `self`.

**generator**: a special type of iterator (and iterable), built using a generator function, i.e. a function with the `yield` keyword.

```python
class MyIterator:
    def __init__(self):
        self.k = 0
    
    def __next__(self):
        if self.k == 10:
            raise StopIteration
        self.k += 1
        return self.k
```

```python
it = MyIterator()
next(it)
```

```python
for k in it:
    print(k)
```

```python
iter(it)
```

```python
class Iterable:
    
    def __iter__(self):
        return MyIterator()
```

```python
for k in Iterable():
    print(k)
```

```python
iter(Iterable())
```

```python
class Iterable2:
    def __init__(self):
        self.k = 0
    
    def __next__(self):
        self.k += 1
        return self.k
    
    def __iter__(self):
        return self
```

```python
for k in Iterable2():
    print(k)
    if k >= 20:
        break
```

```python
# we can do an Iterable by implementing a generator function (=an iterator)
# as the iter method of an iterable!
import torch
class UnitTestDataset(torch.utils.data.IterableDataset):

    in_shape = (2, 3, 4)
    out_shape = (4,)

    def __iter__(self):
        for _ in range(6):
            im = torch.rand(*self.in_shape)
            yield im, 0
# gotchas with pytorch:
# - no random sampling
# - beware multi-gpu trainings with this, you may have the same sequence of data
```

```python
dataset = UnitTestDataset()
for x, y in dataset:
    print(x)
```
