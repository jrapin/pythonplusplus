## Private attributes

```python
class MyClass:
    
    def __init__(self):
        self._attr1 = 1
        self.__attr2 = 12
    
    def print1(self):
        print(f"attributes have values {self._attr1} and {self.__attr2}")
        
class MySubclass(MyClass):

    def __init__(self):
        super().__init__()
        self._attr1 = 2
        self.__attr2 = 24
    
    def print2(self):
        print(f"attributes have values {self._attr1} and {self.__attr2}")
```

```python
obj1 = MyClass()
obj2 = MySubclass()
```

## Generators
Generators don't always behave as you would expect
More details here: https://github.com/satwikkansal/wtfpython#-evaluation-time-discrepancy

```python
# What do you expect?
array = [1, 8, 15]
# A typical generator expression
gen = (x for x in array if x in array)
array = [2, 8, 22]
print(list(gen))
```

```python
# What do you expect?
array_1 = [1, 2, 3]
array_2 = [10, 20, 30]
gen = (i + j for i in array_1 for j in array_2)

array_1 = [4, 5, 6]
array_2 = [400, 500, 600]
print(list(gen))
```

```python
# What do you expect?
powers_of_x = [lambda x: x**i for i in range(10)]
print([f(2) for f in powers_of_x])
```

```python
# new version
funcs = []
for x in range(7):
    def some_func():  # some_func(x=x)
        return x
    funcs.append(some_func)
print([func() for func in funcs])
```

## For-else
Did you know there is an else clause in for/while loops???

```python
for k in range(10):
    print(k)
else:
    print("Else clause") 
print("Finished")
```

```python
for k in range(10):
    print(k)
    if k == 5:
        break
else:
    raise RuntimeError("Did not find k=5") 
print("Finished")
```
