```python
import cProfile
import time
import numpy as np
# import numpy.random
pr = cProfile.Profile()

def myfunc():
    time.sleep(0.1)
    x = np.random.rand(1024, 1024)
    y = x.dot(x)
    print(y)
    
with pr:
    myfunc()
    

pr.dump_stats("slow.prof")
```
