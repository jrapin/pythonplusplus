# Standard library

### collections

```python
import collections
mydict = collections.defaultdict(list)
```

```python
mydict["something"].append(12)
print(mydict)
```

```python
# then again, for containers you can avoid it
normaldict = {}
normaldict.setdefault("something", []).append(12)
print(normaldict)
```

```python
counter = collections.Counter("whatever is iterable")
print(counter)
```

Others:
- `ChainMap`: a dict-like object aggregating several dictionaries
- `deque`: like a list, but fast to operate from both ends, and you can set a maximum size (appending on one hand discards elements of the other end).

### itertools

My whole time favorite!
You should definitely [read the whole doc](https://docs.python.org/3/library/itertools.html)
including (and above all) the "recipes" at the end!

**Remember**: these are all iterators, the elements are fetched on the fly

```python
import itertools
list(itertools.chain.from_iterable([range(2), "abc"]))
```

```python
list(itertools.dropwhile(lambda x: x<5, [1,4,6,4,1]))
```

```python
list(itertools.product("AB", range(3)))
```

```python
list(itertools.combinations("ABCD", 3))
```

```python
list(itertools.permutations("ABCD", 4))
```

```python
list(zip("ABC", range(4)))
```

```python
list(itertools.zip_longest("ABC", range(4), fillvalue="whatever"))
```

### functools
Many cool tools for functions: 
https://docs.python.org/3/library/functools.html

```python
import functools

# @functools.lru_cache
def fibonacci(n: int) -> int:
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

fibonacci(35)
```

```python
def my_func(x: int, y: int) -> int:
    return x + y * 100
```

```python
prefilled_func = functools.partial(my_func, 12)
```

```python
prefilled_func(1)
```

### concurrent.futures
Multithreading/multiprocessing made easy (served as basis for `submitit` syntax)

```python
import concurrent.futures
import operator

jobs = []
with concurrent.futures.ThreadPoolExecutor(max_workers=4) as exc:
    # also available: ProcessPoolExecutor
    for k in range(12):
        jobs.append(exc.submit(operator.add, 12, k))
```

```python
jobs
```

```python
[job.result() for job in jobs]
```

### dataclass

```python
import dataclasses

@dataclasses.dataclass
class MyStruct:
    a: int
    b: str = "a string"
```

```python
MyStruct()
```

```python
MyStruct(12)
```

- fast to define
- simple to use
- comes with extractors such as nice representation, can be made non-mutable etc...

**Note**: always be careful with mutable objects (true everywhere)

```python
import typing as tp

@dataclasses.dataclass
class MyStruct:
    a: int
    b: str = "a string"
    mutable: list[int] = dataclasses.field(default_factory=list)
```

### pathlib

Absolutely essential, if you don't know it yet, [go check it](https://docs.python.org/3/library/pathlib.html) out and never use `os.path` anymore

```python
from pathlib import Path
fp = Path("test.txt")
```

```python
print(f"{fp.absolute() = }")
print(f"{fp.exists() = }")
```

```python
print(fp.absolute().parent.with_name('stuff') / 'whatever.txt')
```

```python
with fp.open("w") as f:
    f.write("12")
```

```python
print(f"{fp.exists() = }")
print(f"{fp.read_text() = }")
```

### tempfile
Create a temporary file/directory

```python
import tempfile
with tempfile.TemporaryDirectory() as tmp:
    fp = Path(tmp) / "somefile.txt"
    fp.touch()
    print(fp)
    print(f"{fp.exists() = }")
```

```python
print(f"{fp.exists() = }")
```
